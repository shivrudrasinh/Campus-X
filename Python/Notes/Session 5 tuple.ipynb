### Difference between lists and tuples 

1) Syntax  2) Mutablity - Tuples are immutable and list is mutable   3) Speed - Tuple are faster the list, Generally immutable datatype are faster compare to mutable datatype   
4) Memory -  5) Build in functionality - are more in list  6) Error prone - list is more error prone , tuple is better 
7) Usablity 

```python
# Speed 

import time 

l = list(range(1000000000))
t = tuple(range(1000000000))

start = time.time()
for i in l:
    i*5
print('List time',time.time() - start)

start = time.time()
for i in t:
    i*5
print("Tuple time",time.time() - start)    
```

```python
# Space

import sys 

l = list(range(1000))
t = tuple (range(1000))

print("list size", sys.getsizeof(l))
print("tuple size", sys.getsizeof(t))
```

## Sets 

#### A set is an unordered collection of items. Every set element is unique(no duplicate) and must be immutable(cannot be changed) however, a set itself is mutable. We can add or remove items from it.

set can also be used to perform mathematical set operations like union,itersection, symmetric difference,etc  

Chracteristics: 
1) Unordered 2)Mutable 3) No duplicate 4) Can't contain mutable data types - Set itself is immutable but it can not contain mutable data types in it 

#### Creating Sets 

```python
# empty 

s = {}  # This is a dictionary so remember  
print(s)
print(type(s))

  # Correct way to create a empty set

a = set()
print(a)
print(type(a))




# 1D and 2D

s1 ={1,2,3}
print(s1)   # This is 1D set / homo set as all values are of same data type 

# 2D Set is not allowed 



# Homo and hetro 

s3 = {1,2.4,"Hello",True,(1,2,3)}  # Python treats 1 are True so 1 comes only 1 time , You can also 
                           # add tuple in it as they are unmuttable  

print(s3) # The items comes in random as they are not ordered , the position is decided by hashing 


# Using type conversion

s4 = set([1,2,3])
print(s4)

# Duplicates not allowed

s5 = {1,1,1,2,3,3,4,4}
print(s5) # All item will appera only one time 


# Set can't have mutable items 

s6 = {1,2[3,4]}
print(s6)
```

```python
s1 = {1,2,3}
s2 = {3,2,1}

print(s1 == s2)
```

### Adding items 

```python
s = {1,2,3,4}

# Add - add single value at a time 
s.add(5)
print(s)


#Update - adds multiple value at a time 
s.update([5,6,7])  # make a list and then add numbers 
print(s)
```

### Deleting items 

```python
# del - delete whole set , cannot delete a particular item as indexing does not work
 
s = {1,2,3,4,5,6,7}
print(s)
del s
print(s)



# discard - you specify which item to delete, if item not present and you discard it then
            # It will not give error  

s.discard(5)
print(s)


# remove - It is same as discard,but  if item not present and you discard it then
            # It will give error   

s.remove(50)

# pop - it randomly delete any item 

s.pop()


# clear - it empty the set 

s.clear()
print(s) 
```

### Set Operation

```python
s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

# Union(|) - prints all the items of the sets once 

s1 | s2

# Intersection(&) - It prints common of both the sets
s1 & s2

# Difference(-) - Such item of A not in B 
s1 - s2 
s2 - s1

# Symmetric Diffrence(^) - except common everyother item 
s1 ^ s2

# Membership test 
1 in s1 
1 not in s1 

# Iteration 
for i in s1:
    print (i)

```

### Set Funtions 

```python
# len / sum / min / max / sorted 

s = {3,1,4,5,2,7}

len(s)
sum(s)
min(s)
max(s)
sorted(s)
sorted(s,reverse=True)
```

```python
# Union/ Update 

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.union(s2)      # What it does Combines elements of both sets 
                  # Removes duplicates Returns a new set Does NOT change s1 

s1.update(s2)    # What it does:Adds all elements of s2 into s1 Changes the original set
                 # permanently 
print(s1)
print(s2)
print(s1)
```

```python
# intersection / intersection_update 

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.intersection(s2)

s1.intersection_update(s2) # What it does: This is the "destructive" version.
                           # It calculates the intersection but instead of returning a new set,
                           # it modifies the original s1 to keep only the overlapping elements.
print(s1)
print(s2)
print(s1)
```

```python
# Difference / Difference Update 

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.difference(s2)

s1.difference_update(s2) 

print(s1)
print(s2)
print(s1)
```

```python
# Symmetric_difference / symmetric_difference_update 

s1 = {1,2,3,4,5}
s2 = {4,5,6,7,8}

s1.symmetric_difference(s2) # It creates a new set containing only the elements that are unique to
                            # each set.

s1.symmetric_difference_update(s2)  # to keep only the unique elements from both sets. 

print(s1)
print(s1)
```

```python
# isdisjoint - The Question: "Do these two sets have nothing in common?"

s1 = {1,2,3,4}
s2 = {4,5,6}

s1.isdisjoint(s2)



#   issubset - The Question: "Are all elements of my set inside the other set?"
s1.issubset(s2)


#  issuperset -The Question: "Does my set contain everything that is in the other set?
s1.issuperset(s2)

```

```python
# Copy  - is used to create a shallow copy of a set.

s1 = {1,2,3}
s2 = s1.copy()

print(s1)
print(s2)
```

## Dictionary Comprehension  

### {key : value for variables in iterable}

```python
# print 1st 10 numbers and their squares 

{i : i**2 for i in range(1,11)}
```

```python
#Using existing dict 

distances = {"delhi":100,"mumbai" :40,"Ap": 43}   # converting values given in km to miles 
{key:value*0.62 for (key,value) in distances.items()}
```

```python
# Using zip - It combines to list to form a dict 
days = ["sunday",'monday','tue']
temp = [31,23,54]

{i:j for (i,j) in zip(days,temp)}
```

```python
# Using if condition 

products = {'phone':10,'laptop':0,'charger':32}  

{ key:value for (key,value) in products.items() if value >0 }
```

```python
# Nested Comprehension 
# Print tables of number from 2 to 4 

{i:{j:i*j for j in range(1,11)} for i in range(2,5)}
```
