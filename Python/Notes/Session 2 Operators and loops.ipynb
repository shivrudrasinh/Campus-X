 # Operator in Python 

~ Arithmetic operator

~ Relational operator

~ Logical operator

~ Bitwise operator

~ Assignment operator

~ Membership operator 

```python
# Arithmetic Operator 
print(5+6)
print(5-6)
print(5*6)
print(5/6)
print(5%6)
print(5//6)
print(5**2)
```

```python
# Relational Operator 

print(4>5)
print(4<5)
print(4<=10)
print(10>=11)
print(4==4)
print(5!=10)
```

```python
# Logical Operator

print(1 and 0)
print(1 or 0)
print(not 1)
```

```python
# Bitwise Operator 

# Bitwise and
print(2 & 3)

# Bitwise or
print(2 | 3)

# Bitwise xor
print(2^3)

# Bitwise not
print(~3) 

#Bitwise left
print (4>>2)

# Bitwise right
print(4<<2)

```

```python
# Assignment operators 
a = 2
a += 2  # This line means a = a + 2  

  
```

```python
# Membership operator

# In
print('d'in 'delhi')

# not in 
print('a' not in 'abad')
```

```python
# Program - Find the sum of 3 digit number entered by the user 

num = int(input("Enter a 3 digit number: "))
print(num)

hundereds = num // 100
tens = (num // 10) % 10
units = num % 10

sum = hundereds + tens + units

print("The sum of three digits is:",sum)


```

```python
# Find the minimum of three numbers 

a = int(input("Enter first number:"))
b = int(input("Enter second number:"))  
c = int(input("Enter third number:"))

if a>b and a>c:
    print('a is greatest',a)

elif b>a and b>c:
    print('b is greatest:',b)

else: 
    print('c is greatest:',c)    
```

```python
# Menu driven calculator

fnum = int(input("Enter the first number:"))
snum = int(input("Enter the first number:"))

op = input("Enter the operation ")

if op =='+':
    print(fnum + snum)

elif op =='-':
    print(fnum - snum)

elif op =='*':
    print(fnum - snum)

elif op =='/':
    print(fnum - snum)

else:
    print("Invalid operation")        


```

```python
# While loop example -> Program to print the table 
num = int(input("Enter number:"))
i = 1
while i<=10:
    print(f"{num} X {i} = {num*i}")
    i +=1
```

```python
# while loop with else

i =1

while i<=5:
    print(i)
    i +=1
else:
    print("Limit exceded")
```

```python
# Guessing game 

import random 
num = random.randint(1,3)
guess = int(input("Enter number between 1 to 10:"))
if guess == num:
    print("Congratulations! You guessed it right.")

elif guess < num:
    print("Try again! Your guess is too low.")

elif guess > num:
    print("Try again! Your guess is too high.")

else:
    print("Invalid input")   


                                       # OR 
                                    

import random 
num = random.randint(1,3)
guess = int(input("Enter a number:"))
counter = 1
while guess != num:
    if guess > num:
        print("You are guessing too high")

    else:
        print("You are guessing too low")

    guess = int(input("Enter a number again:"))       

    counter +=1
    print("attempts",counter)

else:
    print("You have guessed it correct")   
    print("You took",counter,"attempts to guess the number")     
```

# For loop

```python
for i in range(1,11):
    print(i)

for s in range(9,0,-1):
    print(s)    
```

```python
# For loop example - Program - The current population of a town is 1000. Every year the population increases 
                              # by 10% per year. Find the population at the end of each of the last 10 years.


population = 1000
rate = 1.10
for year in range(10,0,-1):
    population = int(population // rate)
    print(f"Population of year {year} is {population}")


```
