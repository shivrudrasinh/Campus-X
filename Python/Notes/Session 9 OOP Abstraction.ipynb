### Aggregation 

```python
# You cannot use private attributes when using aggregation 
# if we want to display private attribute in output then use getter to do that 


'''
You have a person and a car.

A person can own a car.

Person  ------->  Car
          has a

Now think:

If the person dies, does the car disappear? ❌ No.
If the car is sold, does the person disappear? ❌ No.

They both can exist independently.

This is called Aggregation.

'''

class Customer:

    def __init__(self,name,gender,address):
        self.name = name 
        self.gender = gender 
        self.address = address  


    def print_address(self):
        print(self.address.get_city(),self.address.pin,self.address.state)


    def edit_profile(self,new_name,new_city,new_pin,new_state): 
        self.name = new_name
        self.address.edit_address(new_city,new_pin,new_state) 



class Address :

    def __init__ (self,city,pin,state):
        self.__city = city
        self.pin = pin 
        self.state = state 


    def get_city(self):
        return self.__city  


    def edit_address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state


add1 = Address("gurgoan",1221,"haryana")
cust = Customer("nitish","male",add1)


cust.print_address()
cust.edit_profile("ankit","mumbai",111,"maharastra")
cust.print_address()
```

### Super Keyword 

```python

class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price 
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    
    def buy(self):
        print("Buying a smartphone")
        super().buy()  # ~ Syntax to call parent ka buy method 


s = SmartPhone(200,"Apple",13)
s.buy()     
```

```python
# Super -> constuctor 

# You cannot use super keyword outside the class, it will always be use inside the child class 
# You cannot access attributes of parents class using super you can only access methods 


class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.price = price 
        self.brand = brand 
        self.camera = camera 

    

class SmartPhone(Phone):

    def __init__(self,price,brand,camera,os,ram):
        super().__init__(price,brand,camera)
        self.ram = ram 
        self.os = os
        print("Inside Smartphone constructor")


s = SmartPhone(200,"MAC",12,"And",2)

print(s.os)
print(s.brand)
```

### Inheritance summary 

~ A class can inherit from another class.

~ Inheritance improves code reuse.

~ Constructor, attributes,methods get inherited to the child class

~ The parent has no access to the child class

~ Private properties of parents are not accessible directly in child class

~ Child class can override the attributes or methods. This is called method overriding 

~ Super() is an inbuilt function which is used to invoke the parent class methods and constructor

```python
# Diamond Problem

"The Diamond Problem occurs in multiple inheritance, when a class inherits from two classes that both inherit from the same parent"


class Phone:

    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class Product:

    def buy(self):
        print("Product buy method")


# Method resolution order 
class SmartPhone(Product, Phone):   # If two class has same method then the first pass in parameter is called then second 
    pass

s = SmartPhone(200,"apple",13)
s.buy()

```

```python
# Questions 


class A:

    def m1(self):         # m1() -> 20
        return 20
    
class B(A):

    def m1(self):       # m1() -> 30   (overridden)
      return 30         # m2() -> 40
      
    
    def m2(self):
        return 40
    
class C(B):

    def m2(self):              #   m1() -> inherited from B -> 30
        return 20              #   m2() -> overridden in C -> 20
        
    
obj1 = A()    
obj2 = B()    
obj3 = C()

print(obj1.m1() + obj3.m1() + obj3.m2())


'''
When you call a method, Python searches in this order:

Object
   ↓
Current Class
   ↓
Parent Class
   ↓
Grandparent Class
   ↓


For obj3.m1():

obj3
 ↓
Class C   ❌ m1 not found
 ↓
Class B   ✅ found (returns 30)
 ↓
Stop searching   


'''
```

```python
class A:

    def m1(self):
        return 20
    
class B(A):

    def m1(self):
        val = super().m1() + 30
        return val

class C(B):

    def m1(self):
        val = self.m1() + 20
        return val

obj = C()
print(obj.m1())        

```

```python
# Method Overriding 


""" When a child class provides its own implementation of a method that already exists in the parent class, 
    it is called method overriding """

class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


obj = Dog()
obj.sound()

# Python first looks in the child class. Since Dog has its own sound(), it never calls the parent's version.
```

```python
# Method Overloading 

# Method overloading means having multiple methods with the same name but different parameters.

'''No. Python does not support traditional method overloading like Java or C++.

In Python, if you define multiple methods with the same name in a class, the last definition replaces all the previous ones.'''

class Demo:

    def show(self, a):
        print("One argument")

    def show(self, a, b):
        print("Two arguments")


obj = Demo()
obj.show(10,20)



# How do Python programmers achieve similar behavior?

# 1. Using default arguments


class Demo:

    def show(self, a, b=None):
        if b is None:
            print("One argument:", a)
        else:
            print("Two arguments:", a, b)


obj = Demo()

obj.show(10)
obj.show(10, 20)

'''Output:

One argument: 10
Two arguments: 10 20 '''


# 2. Using *args


class Demo:

    def show(self, *args):
        print("Arguments:", args)


obj = Demo()

obj.show(10)
obj.show(10, 20)
obj.show(10, 20, 30)


'''Output:

Arguments: (10,)
Arguments: (10, 20)
Arguments: (10, 20, 30) '''


```

```python

# Operator Overloading 

# Operator overloading means giving operators (+, -, *, ==, >, etc.) a special meaning for objects of your own class.
# Python does this using special methods (magic methods or dunder methods).

# Same operator performs different on basis of the output 

'''
| Operator | Magic Method    |
| -------- | --------------- |
| `+`      | `__add__()`     |
| `-`      | `__sub__()`     |
| `*`      | `__mul__()`     |
| `/`      | `__truediv__()` |
| `==`     | `__eq__()`      |
| `>`      | `__gt__()`      |
| `<`      | `__lt__()`      |
| `>=`     | `__ge__()`      |
| `<=`     | `__le__()`      |
| `!=`     | `__ne__()`      |


'''

class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x,
                     self.y + other.y)

    def display(self):
        print(self.x, self.y)


p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2

p3.display()
```

### Abstraction

```python
# Abstraction is one of the four main principles of Object-Oriented Programming (OOP). It means hiding the internal implementation
#  details and showing only the essential functionality to the user.


# This is a Abstract class, a class such that it has atleast one abstract method in it,(abstract method has no line of code in it)
# You cannot make object of abstract class 

from abc import ABC,abstractmethod
class BankApp(ABC):

    def database(self):
        print("connected to database")


    @abstractmethod       # @abstractmethod is use to make a method abstract 
    def security(self):   # abstract method 
        pass    

```

```python
class MobileApp(BankApp):


   def mobile_login(self):
      print("Login into mobile")

   def security(self):
      print("Mobile security")   
```

```python
mob = MobileApp()
mob.database()
mob.security()
```
