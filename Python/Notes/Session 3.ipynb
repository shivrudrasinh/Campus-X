# Nested loops

```python
# Examples -> Unique pairs

for i in range(1,5):
    for j in range(1,5):
        print(i,j)
        
```

```python
# Pattern 1 

# *
# **
# ***

row = int(input("Enter number of rows:"))

for i in range(1,row+1):
    for j in range(1,i+1):
     print("*",end="")
    print()
```

```python
# Pattern 2

''' 
1
121
12321
1234321
'''

rows = int(input("Enter number of rows"))

for i in range (1,row+1):
    for j in range(1,i+1):
        print(j,end="")
    for k in range(i-1,0,-1):
        print(k,end='')

    print()       
 
```

# Loop control Statement 

-> Break


-> Continue


-> Pass

```python
# Break

for i in range(1,10):
    if i ==5:
        break
    print(i)
```

```python
start = int(input("Enter starting number:"))
end = int(input("Enter ending number:"))

for i in range(start,end+1):
    for j in range(2,i):
     if i%j ==0:
        break
    else:
        print(i)
```

```python
# Continue 

for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
# Pass

for i in range(1,17):
    pass

# It is just to prevent errors
```

# Accessing substrings from a String 

```python
# Positive Indexing 

s = " Hello World"
print(s[5])

# Negative Indexing
print(s[-1])
```

```python
# Slicing  

# In reverse slicing the starting number is always greater than end number 

z= "Hiroshima Nagasaki"
print(z[0:7:2])
print(z[:7:2])
print(z[6:0:-1])

# Reverse String 
print(z[::-1])
```
