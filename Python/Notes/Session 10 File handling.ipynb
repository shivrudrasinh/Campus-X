```python
# To read line by line  -> readline()


f = open("sample.txt","r")
print(f.readline(),end="") 
print(f.readline(),end="")            # As many time you write readline that many time lines will execute 
f.close()                      # write one time then first line will be read , second time then second line will print
                              
```

### Using context manager (With)

~ It's a good idea to close a file usage as it will free up the resources 

~ If we don't close it,garbage collector would close it

~ with keyword closes the file as soon as the usage is over 

```python
# with

with open("sample.txt","w") as f:
    f.write("Veer bhogya vasundhara")
```

```python
# try f.read() now 

with open("sample.txt","r") as f:
    print(f.read())
```

```python
# Moving within a file -> 10 char then 10 char 

with open("sample.txt","r") as f:
    print(f.read(10))  # This will print first 10 character
    print(f.read(10))  # This will print next 10 character 

    # This method is use when the file is to big to load in memory eg 10gb then we can first load some part then other so on 
```

```python
# benefit ?  -> To load a big file in memory 

big_l = ['hello world ' for i in range(1000)]

with open("big.txt","w") as f:
    f.writelines(big_l)
```

```python
with open("big.txt","r") as f:

    chunk_size = 10

    while len(f.read(chunk_size)) > 0:
        print(f.read(chunk_size),end="")
        f.read(chunk_size)
```

```python
# seek and tell function

with open("sample.txt","r") as f:
    print(f.read(10))  # This will print first 10 character
    print(f.tell())   # This will print the current position of file pointer 

    print(f.seek(0))  # This will move the file pointer to the start of the file
    print(f.tell())  # The pointer is on 0 now again 
    print(f.read(10))
```

```python
# seek during write 

with open("sample.txt","w") as f:
    f.write("hello")
    f.seek(0)
    f.write('x')
```

### Serialization and Desearialization 

~ Serialization : process of converting python data types to JSON format 

~ Deserialization : process of converting JSON to python data types 

```python
# JSON : Javascript on notation  

# Serialization using json module 
# dump : it take 2 input (what to input , file handler object )

# List 
import json
l = [1,2,3,4]

with open("demo.json","w") as f:
    json.dump(l,f)
```

```python
# dict

import json
d = {

    'name' : "nitish",
    'age' : 33,
    'gender' : 'male'
}

with open("demo.json","w") as f:
    json.dump(d,f,indent=4)        # indent is use for spacing
```

```python
# Deserialization 

# load take 1 input file handler name 

import json 

with open("demo.json","r") as f:
   d =  json.load(f)
   print(d)
   print(type(d))
```

```python
import json 

t = (1,2,3,4,5)
with open("demo.json","w") as f:
    json.dump(t,f)
```

```python
# serialize and deserialize a nested dict 

d = {
    "student": "nitish",
    "marks" : [22,33,32,12]

}

with open("demo.json","w") as f:
    json.dump(d,f)
```

### Serialization and deserializating custom objects 

```python
class Person:

    def __init__(self,fname,lname,age,gender):
        self.fname = fname 
        self.lname = lname
        self.age = age
        self.gender = gender

# format to printed in 
# Nitish singh age -> 33 gender  -> male        
```

```python
person = Person("nitish", "singh",33,"male") 
```

```python
# As a string

import json

def show_object(person):
    if isinstance(person,Person):
      return"{} {} age -> {} gender -> {}".format(person.fname,person.lname,person.age,person.gender)

with open('demo.json',"w") as f :
    json.dump(person,f,default=show_object,indent=4)
```

```python
# As dict 

import json

def show_object(person):
    if isinstance(person,Person):
      return { "name" : person.fname + ' ' + person.lname, "age" : person.age, "gender" : person.gender}

with open('demo.json',"w") as f :
    json.dump(person,f,default=show_object,indent=4)
```

```python
# Deserializating

import json 

with open("demo.json","r") as f:
    print(json.load(f))
```
