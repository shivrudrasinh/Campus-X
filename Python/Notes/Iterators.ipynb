### What is an Iteration

Iteration is a general term for taking each item of something, one after another. Any time you use
a loop, explicit or implicit, to go over a group of items, that is iteration. 

### What is Iterator

An Iterator is an object that allows the programmer to traverse through a sequence of data without having to store the
entire data in the memory

```python
# Example 


import sys

l = [ x for x in range(1,100)]

for i in l:
    print(i*2)
sys.getsizeof(l)/1024


x = range(1,100)

for i in x:                # This takes much less memory compare to above 
    print(i*2)

print(sys.getsizeof(x)/64)
```

### What is Iterable

Iterable is an object, which one can iterate over

It generates an iterator when passes to iter() method

```python
l = [1,2,3]
type(l)

# L is an iterable
type(iter(l))

# iter(l)  --> iterator
```

### Point to remember
