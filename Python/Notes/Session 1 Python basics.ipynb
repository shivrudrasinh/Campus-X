5. Type Conversion 

```python
# There are two types of conversion 
# 1. Implicit    2. Explicit  
 
#  Implicit is done internally by the compiler itself 
print(5+6.0)
 
 

# and  explicit is done manually 
print(int("4"))
```

```python
# In python when explicit conversion is perform it doesnot change the data type of variable instead it creates new 
# variable with the changed datatype So, the prior variable datatype remains the same this is only applied when 
#  the type conversion in not perform will taking input and storing value in variable  

ty= "10" 

print(type(ty))
sy = 1
print(int(ty)+ sy)
```
