### Exceptions

If things go wrong during the executiob of the program(runtime). It generally happens when something unforeseen has happened.

~ Exceptions are raised by python runtime

~ You have to takle is on fly

Examples :

~ Memory overflow

~ Divide by 0 -> logical error

~ Database error


```python
# Why is it important to handle them
# How to handle exceptions
# -> Try except block 
```

```python
# try catch demo

try:
    with open("sample1.txt","r") as f:
        print(f.read())
except:
    print("Sorry file not found")
```
