```python
# Syntax to crete an object

#  objectname = classname()


class Atm:

    # Constructor  : it is a special function, you donot need to need to class the function to run it happens by itself when a object of
    # class is created

    def __init__(self):
        self.pin = ""
        self.balance = 0
        print("Me to execute ho gaya")
        self.menu()

    def menu(self):
        user_input = input(""""
        Hi how can I help you?
        1. Press 1 to create pin 
        2. Press 2 to change pin 
        3. Press 3 to check balance 
        4. Press 4 to withdraw 
        5. Anything else to exit  
               
              """)

        if user_input == "1":
            # create pin
            self.create_pin()

        elif user_input == "2":
            # change pin
            self.change_pin()

        elif user_input == "3":
            # check balance
            self.check_balance()

        elif user_input == "4":
            self.withdraw()
            pass

        else:
            exit()

    def create_pin(self):
        user_pin = input("enter your pin")
        self.pin = user_pin

        user_balance = int(input("Enter your balance"))
        self.balance = user_balance

        print("Pin created successfully")
        self.menu()

    def change_pin(self):
        old_pin = input("Enter old pin")

        if old_pin == self.pin:
            # Let him change the pin
            new_pin = input("Enter new pin")
            self.pin = new_pin
            print("Pin change successful")
            self.menu

        else:
            print("Nahi karne de sakta re baba")
            self.menu()

    def check_balance(self):
        user_pin = input("Enter your pin")
        if user_pin == self.pin:
            print("Your balance is", self.balance)
        else:
            print("Chal Nikal yaha se ")
        self.menu()

    def withdraw(self):
        user_pin = input("Enter your pin")
        if user_pin == self.pin:
            # allow to withdraw
            amount = int(input("Enter amount to withdraw"))
            if amount <= self.balance:
                self.balance = self.balance - amount
                print("Withdraw successfully, Balance is", self.balance)

            else:
                print("Abe gareeb")

        else:
            print("Kal ana")

        self.menu()
```

```python
# Creating own datatype 

class fraction:
    

    # Parameterized constructor : A constructor that need inputs 
    def __init__(self,x,y):
        self.num = x
        self.den = y


    def __str__(self):
        return '{}/{}'.format(self.num,self.den) 


    def __add__(self, other):  # It automatically trigers when you write + operator 
          
          new_num = self.num*other.den + other.num * self.den 
          new_den = self.den*other.den

          return '{}/{}' .format(new_num,new_den)
    


    def __sub__(self, other):  # It automatically trigers when you write - operator 
          
          new_num = self.num*other.den - other.num * self.den 
          new_den = self.den*other.den

          return '{}/{}' .format(new_num,new_den)
    


    def __mul__(self, other):  # It automatically trigers when you write * operator 
          
          new_num = self.num*other.num 
          new_den = self.den*other.den

          return '{}/{}' .format(new_num,new_den)
    

    
    def __truediv__(self, other):  # It automatically trigers when you write * operator 
          
          new_num = self.num*other.den
          new_den = self.den*other.num

          return '{}/{}' .format(new_num,new_den)
          
          
    def convert_to_decimal(self):
         return self.num / self.den 
```

```python
fr1 = fraction(3,4)
fr2 = fraction(2,3)
```

```python
print(fr1)  # Whenever you want to print something in class python goes into the _str_ magic method 
print(fr2)
print(fr1 + fr2)
print(fr1 - fr2)
print(fr1 * fr2)
print(fr1 / fr2)
fr1.convert_to_decimal()
```
