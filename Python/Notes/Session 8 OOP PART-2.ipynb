### Reference Variable 

-> Reference variable hold the objects


-> We can create objects without reference variable as well


-> An object can have multiple reference variable 


-> Assigning a new reference variable to an existing object does not create a new object  

```python
# object without a reference 

class Person():

    def __init__(self):
        self.name = 'nitish' 
        self.gender = 'male'


p = Person()     # Here when we call Person() and object was created in memory and memory address of that is store 
                 # in p    " p is not the object it is a reference variable that contains the address " 
q = p 
```

```python
# Multiple ref 

print(id(p))
print(id(q))
print(p.name)
print(q.name)
q.name = 'ankit'
print(p.name)
print(q.name)
```

### Pass by reference

```python
# An object can be given as input to a function 
# An function can given in return a object of a class 



class Person:

    def __init__(self,name,gender):
        self.name = name
        self.gender = gender



# outside the class -> function 

def greet(person):
    print('hi my name is',person.name,'and i am a',person.gender)
    p1 = Person('ankit' , 'male')
    return p1


p = Person("nitish", 'male')    
greet(p)
x = greet(p)
print(x.gender)
```

### Getter and Setter

```python
'''
A getter is a method used to retrieve (get) the value of an object's attribute.

A setter is a method used to modify (set) the value of an object's attribute.

They are mainly used for encapsulation, which means controlling how the data inside an object is accessed or modified. '''

class Student :

    def __init__(self,name,age):
        self.name = name
        self.__age = age   # Private attribute


    # Getter
    def get_age(self):
        return self.__age
 

    # Setter 

    def set_age(self,age):
      
      if age > 0 :
          self.__age = age 
      else:
          print("Age cannot be negative") 


s1 = Student("sHIV",21)

print(s1.get_age())

s1.set_age(22)
print(s1.get_age())

s1.set_age(-4)
print(s1.get_age())




```

### Collection of object

```python

# List of object 


class Person:

    def __init__(self,name,age):
        self.name = name 
        self.age = age 

p1 = Person('nitish','32')
p2 = Person('rana','32')   
# print(p1.name , p1.age)     

l = [p1,p2]   # This means a list can be created so you can perform all the operation that you perform on a list

for i in l:
    print(i.name,i.age)
```

### Dict of object

```python
class Person:

    def __init__(self,name,age):
        self.name = name 
        self.age = age 

p1 = Person('nitish','32')
p2 = Person('rana','32')   


d = { "p1": p1 , "p2" : p2  }

for i in d:
    print(i)    # Printing i will give all the keys 
    print(d[i])
    print (d[i].name,d[i].age )
```

### Static variable  vs  Instance variable 

```python
# Instance Variable → Each object has its own copy.
# Static (Class) Variable → One copy is shared by all objects.
```

```python
# Instance varialbe -


#  An instance variable belongs to an object (instance).
# It is created inside the __init__() method using self.


class Person:

    def __init__(self,name,gender):
        self.name = name             # instance variable
        self.gender = gender         # instance variable 


p1 = Person("rana",32)
p2 = Person("shiv",21)

print(p1.name,p1.gender)
print(p2.name,p2.gender)
```
