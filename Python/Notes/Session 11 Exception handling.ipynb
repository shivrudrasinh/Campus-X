### Exceptions

If things go wrong during the executiob of the program(runtime). It generally happens when something unforeseen has happened.

~ Exceptions are raised by python runtime

~ You have to takle is on fly

Examples :

~ Memory overflow

~ Divide by 0 -> logical error

~ Database error


```python
# Why is it important to handle them
# How to handle exceptions
# -> Try except block 
```

```python
# Lets create a file

with open("sample.txt","w") as f:
    f.write("hello world")
```

```python
# try catch demo

try:
    with open("sample1.txt","r") as f:
        print(f.read())
except:
    print("Sorry file not found")
```

```python
# catching specific exception
try:
    m=5
    f = open('sample.txt','r')
    print(f.read())
    print(m)
    print(5/2)
    l = [1,2,3]
    print(l[100])
except FileNotFoundError:
    print('file not found')

except NameError:
    print('variable not defined')

except ZeroDivisionError:
    print("can't divide by 0")

except Exception as e:     # This block should always come in last otherwise this will print by default 
    print(e)    
```

```python
# else 

try:
    f = open("sample1.txt","r")
except FileNotFoundError:
    print("file nahi he")
except Exception:
    print("Kuch to lafda hai")   
else:
    print(f.read())    
```

```python
# Finally 


try:
    f = open("sample1.txt","r")
except FileNotFoundError:
    print("file nahi he")
except Exception:
    print("Kuch to lafda hai")   
else:
    print(f.read())  
finally:
    print("Ye to print hoga he")    
```

```python
# Raise exception 

# In python programming, exceptions are raised when errors occur at runtime.
# We can also manually raise exceptions using the raise keyword
# We can optionally pass values to the exception to clarify why that exception was raised 

raise FileNotFoundError("Aise he try kar raha hu")
raise NameError("Aise he try kar raha hu")
raise ZeroDivisionError("Aise he try kar raha hu")
raise ModuleNotFoundError("Aise he try kar raha hu")
```

```python
# Creating custom exceptions
# Exception hierarchy in python

class MyException(Exception):
    def __init__(self, message):
        super().__init__(message)


class Bank:

    def __init__(self, balance):
        self.balance = balance

    def withdraw(self, amount):

        if amount < 0:
            raise MyException("Amount cannot be negative")

        if self.balance < amount:
            raise MyException("Paise nahi he tere pass")

        self.balance -= amount


obj = Bank(1000)

try:
    obj.withdraw(500)
except MyException as e:
    print(e)
else:
    print(obj.balance)
```

```python
# Simple example 

class SecurityError(Exception):

    def __init__(self,message):
        print(message)

    def logout(self):
        print("logout")    



class Google:

    def __init__(self,name,email,password,device):
        self.name = name
        self.email = email 
        self.password = password
        self.device = device

    def login(self,email,password,device):
        if device != self.device:
            raise SecurityError("bhai tere toh lag gai")
        if email == self.email and password == self.password:
            print("Welcome") 
        else:
            print("Login error")

obj = Google("nitish","nitish@gmail.com","1234","android")    

try:
    obj.login("nitish@gmail.com","1234","windows")
except SecurityError as e:
    e.logout()
else:
    print(obj.name)  
finally:
    print("Database connection closed")         
```
