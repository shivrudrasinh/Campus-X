### Aggregation 

```python
# You cannot use private attributes when using aggregation 
# if we want to display private attribute in output then use getter to do that 


'''
You have a person and a car.

A person can own a car.

Person  ------->  Car
          has a

Now think:

If the person dies, does the car disappear? ❌ No.
If the car is sold, does the person disappear? ❌ No.

They both can exist independently.

This is called Aggregation.

'''

class Customer:

    def __init__(self,name,gender,address):
        self.name = name 
        self.gender = gender 
        self.address = address  


    def print_address(self):
        print(self.address.get_city(),self.address.pin,self.address.state)


    def edit_profile(self,new_name,new_city,new_pin,new_state): 
        self.name = new_name
        self.address.edit_address(new_city,new_pin,new_state) 



class Address :

    def __init__ (self,city,pin,state):
        self.__city = city
        self.pin = pin 
        self.state = state 


    def get_city(self):
        return self.__city  


    def edit_address(self,new_city,new_pin,new_state):
        self.__city = new_city
        self.pin = new_pin
        self.state = new_state


add1 = Address("gurgoan",1221,"haryana")
cust = Customer("nitish","male",add1)


cust.print_address()
cust.edit_profile("ankit","mumbai",111,"maharastra")
cust.print_address()
```

### Inheritance 

```python
'''
Inheritance = "IS-A" relationship

Inheritance means one class is a special type of another class.

For example:

Dog is an Animal.
Car is a Vehicle.

'''

class User:

    def __init__(self):
        self.name = "nitish"

        


    def login(self):
        print("Login")
        


class Student(User):

    def __init__(self):
        super().__init__()
        self.rollno = 100 

    def enroll(self):
        print("Enroll")


s = Student()
u = User()

print(s.name,s.rollno,end=" ")
print()
s.login()
s.enroll()



# class Instructor(User):    
```

```python
# Constructor example 

''' In inheritence if the child class does not constructor then it excecutes the parents class constructor '''

class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.price = price 
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass


s = SmartPhone(200,"Apple",13)
s.buy()
```

```python
# Constructor example 2 

''' In inheritence if the child class has constructor then it excecutes child class constructor and parent constructor is not call 
    and when the parent constructor is not called the variables inside it  does not get initialize so because of that when child class
    try to access the parent class variable it gives error   '''

class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.price = price 
        self.brand = brand 
        self.camera = camera 

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):

    def __init__(self,os,ram):
        self.ram = ram 
        self.os = os
        print("Inside Smartphone constructor")


s = SmartPhone("MAC",8)
s.buy()
# s.brand          ~This gives error 
```

```python

# Child class cannot access parent class private variable 

class Phone:
    def __init__(self,price,brand,camera):
        print("Inside phone constructor")
        self.__price = price 
        self.brand = brand 
        self.camera = camera 

    def show(self):
        print(self.__price)

class SmartPhone(Phone):
    
    def check(self):
        print(self.brand)
        print(self.camera)
        print(self.__price)


s = SmartPhone(200,"Apple",13)
s.show()
s.check()
```

```python
class Parent:

    def __init__(self,num):
        self.__num = num

    
    # A getter is use to print the private variable 
    
    def get_num(self):
        return self.__num
    
class Child(Parent):

    def show(self):
        print("This is in child class")


son = Child(100)
print(son.get_num())
son.show()
```

```python
class Parent:

    def __init__(self,num):
        self.__num = num

    def get_num(self):
        return self.__num
    
    
class Child(Parent):

    def __init__(self,val,num):
        self.__val = val

    def get_val(self):
        return self.__val


son = Child(100,10)
print("Parent: Num:",son.get_num())
print("Child: Val:",son.get_num())
```
