## *args and **kwargs 

```python
# are special python keywords that are used to pass the 
# the variable length of arguments to a function 
```

```python
# *args 

# In Python, *args is used to pass a variable number of positional arguments to a function.
# allows us to pass a variable number of non-keyword argument to a function.
# It creats a tuple and stores all the value in it and perfrom operation then 
# you can use any word instead of args it will work   


def multiply(*args):
    product = 1

    for i in args:
        product = product *i
    print(args)    
    return product  
```

```python
multiply(1,2,3,3,4,4,5)
```

```python
# **kwargs is used to pass a variable number of keyword arguments to a function.
# Keyword arguments mean that they contain a key-value pair,like a python dictionary 


def display(**kwargs):

    for(key,value) in kwargs.items():
        print(key,"->",value)
```

```python
display(india="delhi",srilanka="colombo",nepal="kathmandu")
```

## Lambda Function 

#### A Lambda function is a small anonymous function, it can take any number of arguments,but can only have one expression 

```python
# Syntax :  lambda a,b: a+b      Here: lambda is keyword , a,b is parameters and a+b is expression 
```

```python
''' In Python, a lambda is a small, anonymous (unnamed) function that can have any number of arguments but only one expression.

Syntax
lambda arguments: expression

It is equivalent to writing a normal function with def, but in a shorter form.

Example 1: Add two numbers


Using def:

def add(a, b):
    return a + b
print(add(3, 5))  # 8


Using lambda:

add = lambda a, b: a + b
print(add(3, 5))  # 8  '''
```

```python
a = lambda x:x**2
a(2)

```

### Diff between lambda vs Normal Function 


1) No Name   


2) Lambda has no return value(infact, returns a function)    


3) lambda is written in 1 line   


4) Not reusable 

Then why use lambda funtions ?
-> They are used with " Higher order function "

```python
# Check if a string has "a"

b = lambda s: 'a' in s
b('hello')
```

```python
# Odd or even 

s = lambda x: 'even' if x%2 ==0 else "odd"
s(4)
```

### Higher order function 

```python
# A function that return a function or recevies a function in input 

# Example 

def square(x):
    return x**2 

def transform(f,l):       # f is not a number or string. It is a function.
    output = []
    for i in l:
        output.append(f(i))     # f(i) means square(i)

    print(output)


l = [1,2,3,4]    

transform(square,l)

```

### There are 3 types of HOF  

1) Map  
    

2) Filter     


3) Reduce 

### Map

```python
# MAP  : It takes two things are input lambda function and list 


# Square the itmes of a list 
list(map(lambda x:x**2,[1,2,3,4,5]))


# odd/even labelling of list items
l = [1,2,3,4]
list(map(lambda x:"even" if x%2 ==0 else "odd",l))


# fetch names from a list of dict 

users = [
    {
        'name' :'rahul',
        'age':45,
        'gender' :"male"
    }
]

list(map(lambda users:users['gender'],users))
```

### Filter 

```python
# number greater than 5 
l = [3,4,5,6,7]

list(filter(lambda x:x>5,l))
```

```python
# fetch fruits starting with 'a'

fruits = ['apple','cherry','pine']
list(filter(lambda x:x.startswith('a'),fruits) )
```

### Reduce

```python
# sum of all item
import functools

functools.reduce(lambda x,y:x+y,[1,2,3,4,5])
```

```python
# find min

functools.reduce(lambda x,y:x if x<y else y,[23,11,24,5])
```
