### Namespaces 


A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces:

· Builtin Namespace

· Global Namespace


. Enclosing Namespace


· Local Namespace

### Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.
The interpreter searches for a name from the inside out, looking in the local, englosing, global, and finally the built-in scope. If the interpreter doesn't find the name in any of these locations, then python raises a NameError exception

```python
# local and global 

a = 2    # Global scope

def temp():
    b = 3     # Local scope 
    print(b)

temp()
print(a)
```

```python
# Local and Global  -> same name 

b = 2    

def temp():
    b = 3    

temp()
print(b)
```

```python
# Local and global  -> editing global

# You cannot edit global in local 

b = 2    

def temp():
    b += 1
    print(b)    

temp()
print(b)
```

```python

b = 2    

def temp():
    global b      # When you add global keyword you can edit global variable in local 
    b += 1
    print(b)    

temp()
print(b)
```

```python
# local and global  -> global created inside local 

def temp():
    global b
    b = 1
    print(b)    

temp()
print(b)
```

```python
# Local and global -> function parameter is local


def temp(z):                  # Parameter of a function is always local 
    print(z)

a = 5
temp(5)
print(a)
```

```python
# Enclosing scope 

def outer():

    def inner():
        print("inner function")
    inner()
    print("outer function")

outer()
print("main program")    
```

```python
# non local keyword

def outer():
    a = 1
    def  inner():
        nonlocal a
        a += 1
        print("inner function",a)
    inner()
    print("outer function",a)
 
outer()
print("main program")  
```

### Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and
returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python

-> Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc

-> User defined decorators that we programmers can create according to our needs

```python
# Python function are first class citizen

def func():
    print("hello")

a = func
a()    

```

```python
def modify(func,num):
    return func(num)

def square(num):
    return num**2

modify(square,2)
```

```python
# Simple example 

def my_decorator(func):

    def wrapper():
        print("*************************")
        func()
        print("**************************")
    return wrapper

def hello():
    print("hello")

def display():
    print("hello nitish")

a = my_decorator(hello)    
a()
b = my_decorator(display)  
b() 
```

```python
# better syntax


def my_decorator(func):

    def wrapper():
        print("*************************")
        func()
        print("**************************")
    return wrapper

@my_decorator
def hello():
    print("hello")

hello()
```

```python
# Anything meaninful

import time 

def timer(func):
    def wrapper():
        start = time.time()
        func()     # In this case, func refers to the hello() function.
        print("Time taken by",func.__name__,time.time() - start,"secs")
    return wrapper

@timer

def hello():
    print("hello world")
    time.sleep(2)

hello()    
```

```python
def sanity_check(data_type):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if type(args[0]) == data_type:      # args is actually tuple
                func(*args)
            else:
                raise TypeError("Ye datatype nahi chaliga")
        return inner_wrapper
    return outer_wrapper

@sanity_check(int)
def square(num):
    print(num**2)

square("h")    
```
